```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator  # The modern simulator
```

```python
# 1. Setup: 2 qubits, 2 classical bits
qc = QuantumCircuit(2, 2)
```

```python
# 2. The Logic:
qc.h(0)        # Put first qubit in superposition (The Flip)
qc.cx(0, 1)    # Entangle them (The Link)
```

```python
# 3. The Measurement:
qc.measure([0, 1], [0, 1])
```

```python
# 4. Running the simulator:
# In 2026, we simply initialize the simulator directly
simulator = AerSimulator()
job = simulator.run(qc, shots=1000)
result = job.result()
```

```python
# 5. Show results:
counts = result.get_counts()
print(f"Results: {counts}")
```
